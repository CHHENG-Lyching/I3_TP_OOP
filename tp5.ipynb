```python
#EX: 4.1. Student Information System
class Student:
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id
    def details(self):
        print("Student Name: ", self.name)
        print("Age : ", self.age)
        print("Student ID : ", self.student_id)
    def is_adult(self):
        if self.age < 0:
            print("Invalid age !")
        elif self.age >= 18:
            a = "Yes"
        else:
            a = "No"
        print("Adult :", a,"\n")
    

s1 = Student("Visal", 22, "ST001")
s2 = Student("Ching", 19, "ST004")
s3 = Student("Kimheng", 8, "ST003")

s1.details()
s1.is_adult()
s2.details()
s2.is_adult()
s3.details()
s3.is_adult()
```

```python
#EX: 4.2. Bank Account

class BankAccount:
    bank_name = "ITC Bank"

    def __init__(self, owner, account_number, balance):
        self.owner = owner
        self.account_number = account_number
        self.balance = balance
    def deposit(self, amount):
        if amount > 0:
            print("Deposit: ", amount)
            self.balance += amount
        else:
            print("Invalid amount !")
    def withdraw(self, amount):
        if amount > self.balance:
            print("Your balance is insufficient !")
        elif amount <= 0:
            print("Amount must be above zero !")
        else:
            print("Withdraw: ", amount)
            self.balance -= amount
    def display_balance(self):
        print("Account Owner: ", self.owner)
        print("Initial Balance:  $", self.balance, "\n")

print("Bank: ", BankAccount.bank_name, "\n")
print("=" * 25)

a1 = BankAccount("Visal", "",1000)
a1.display_balance()
a1.deposit(500)
print("New Balance: $", a1.balance)
a1.withdraw(200)
print("Final Balance: $", a1.balance)

print("=" * 25)

a2 = BankAccount("Ching", "",59.99)
a2.display_balance()
a2.deposit(40.01)
print("New Balance: $", a2.balance)
a2.withdraw(10)
print("Final Balance: $", a2.balance)






            
```

```python
#EX4.3: Rectangle Calculator
class Rectangle:
    def __init__(self, width, length):
        self.width = width
        self.length = length
    def area(self):
        return self.width * self.length
    def perimeter(self):
        return 2 * (self.width + self.length) 
    @staticmethod
    def is_valid(width, length):
        return length > 0 and width > 0
            
#create 3 rectangle objects
if Rectangle.is_valid(2, 8):
    r1 = Rectangle(2, 8)
    print("Rectangle 1")
    print("Length: ", r1.length)
    print("Width: ", r1.width)
    print("Area: ", r1.area())
    print("Perimeter: ", r1.perimeter())

print("=" * 25)

if Rectangle.is_valid(10, 4):
    r2 = Rectangle(10, 4)
    print("Rectangle 2")
    print("Length: ", r2.length)
    print("Width: ", r2.width)
    print("Area: ", r2.area())
    print("Perimeter: ", r2.perimeter())

print("=" * 25)

if Rectangle.is_valid(-2, 4):
    r3 = Rectangle(-2, 4)
    print("Rectangle 3")
    print("Length: ", r3.length)
    print("Width: ", r3.width)
    print("Area: ", r3.area())
    print("Perimeter: ", r3.perimeter())

else:
    print("Invalid rectangle dimensions!")


    
```

```python
# 4.5 Temperature Converter
class Temperature:
    @staticmethod
    def celcius_to_fahrenheit(celsius):
        return (9/5) * celsius + 32
    @staticmethod
    def fehrenheit_to_celsius(fahrenheit):
        return (5/9)  * (fahrenheit - 32)
    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15

c =25
print("Celsius: ", c)
print("Fahrenheit: ", Temperature.celcius_to_fahrenheit(c))
print("Kelvin: ", Temperature.celsius_to_kelvin(c))

print("\nMore:")
for c in [0, 100, 37]:
    f =  Temperature.celcius_to_fahrenheit(c)
    k = Temperature.celsius_to_kelvin(c)
    print(f"{c}°C = {f:.2f}°F = {k:} K ")
```

```python
#EX 4.6: Product Management System
class Product:
    store_name = "ITC Online Store"
    product_count = 0
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
        Product.product_count += 1
    def display_product(self):
        print("Name: ", self.name)
        print("Price: $", self.price)
        print("Quantity: ", self.quantity)
    def total_value(self):
        print("Total Value: $", self.price * self.quantity)
    @classmethod
    def get_product_count(cls):
        return cls.product_count
    @staticmethod
    def is_valid_price(price):
        return price > 0

print("============", Product.store_name, "============")
print("\nProduct 1")

p1 = Product("Laptop", 800, 5)
p1.display_product()
p1.total_value()

print("\nProduct 2")

p2 = Product("Mouse", 15, 20)
p2.display_product()
p2.total_value()

print("\nProduct 3")

p3 = Product("Keyboard", 30, 10)
p3.display_product()
p3.total_value()

print("\nTotal Products: ", Product.get_product_count())

total = (p1.price * p1.quantity) + (p2.price * p2.quantity) + (p3.price * p3.quantity)

print("Total Invertory Value: $", total)

        
```
