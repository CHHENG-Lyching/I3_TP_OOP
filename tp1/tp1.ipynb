```python
#G
print("G")
for i in range(5):
    if i <=2 :
        print(" " * (i - 1) + "*" * (7 - 2 * i))
    else:
        print( "*" * (7 - 2 * i) + " " * (i - 1))

```

```python
#F
print("F")
for i in range(6):
    if i <=2:
        print(" " * (3 - i) + "*" * (2 * i - 1))
    else:
        print(" " * (i - 3) + "*" * (11 - 2 * i))
```

```python
#E
print("E")
for i in range(5):
    if i == 0 or i == 4:
        print(" "*4 + "*"*1)
    elif i == 1 or i == 3:
        print(" "*2 + "*"*3)
    else:
        print("*"*5)
```
